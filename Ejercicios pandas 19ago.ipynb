```python
#un primer Dataframe
#Aqui se crea un dataframe de ejemplo.

import pandas as pd

serie = pd.Series([24000, 28000, 22000])
serie.to_numpy () # array ([24000, 28000, 22000])
print (serie)

df = pd.DataFrame ({
    "Producto": ["Bowl Andino", "Pasta Urbana"],
    "precio": [24000, 28000],
})
type(df["precio"].to_numpy()) # <class 'numpy.ndarray'>
#Otra forma de hacerlo es la siguiente
type(df.precio.to_numpy()) # <class 'numpy.ndarray'>

type (df) # metodo para mostrar el tipo de estructura de datos (arreglo np o DataFrame)

df.shape # muestra cantidad de columnas y filas del arreglo
```
