# 1. Sesión 3 — Variables y tipos de datos

Una **variable** es un nombre que permite almacenar un valor.

El operador `=` asigna un valor a una variable.

```python
sede = "Centro"
pedidos = 18
ticket_promedio = 27500.5
meta_cumplida = False

print (sede)
print(pedidos)
print(ticket_promedio)
print(meta_cumplida)
```

Python reconoce distintos tipos de datos:

- `str`: texto;
- `int`: número entero;
- `float`: número decimal;
- `bool`: valor lógico: `True` o `False`.

`type()` permite consultar el tipo de un valor.

```python
print(type(sede))
print(type(pedidos))
print(type(ticket_promedio))
print(type(meta_cumplida))
```

## Nombres de variables

Un nombre descriptivo facilita la lectura del programa.

En Python se acostumbra usar minúsculas y separar palabras con guion bajo.

```python
precio_unitario = 18500
cantidad_productos = 4
nombre_producto = "Bowl Andino"

print(nombre_producto)
print(precio_unitario)
print(cantidad_productos)
```

## Operaciones aritméticas

Los operadores más frecuentes son:

- `+`: suma;
- `-`: resta;
- `*`: multiplicación;
- `/`: división;
- `//`: división entera;
- `%`: residuo;
- `**`: potencia.

```python
cantidad = 4
precio_unitario = 18500
costo_unitario = 11200

ingreso_bruto = cantidad * precio_unitario
costo_total = cantidad * costo_unitario
margen_bruto = ingreso_bruto - costo_total

print("Ingreso bruto:", ingreso_bruto)
print("Costo total:", costo_total)
print("Margen bruto:", margen_bruto)
```

Los paréntesis ayudan a expresar el orden del cálculo.

```python
margen_porcentual = (margen_bruto / ingreso_bruto) * 100

print("Margen porcentual:", margen_porcentual)
print("Margen redondeado:", round(margen_porcentual, 2))
```

## Conversión de tipos

Los datos recibidos desde formularios, archivos o `input()` suelen llegar como texto.

Para calcular con ellos es necesario convertirlos.

```python
cantidad_texto = "4"
precio_texto = "18500.50"

cantidad_convertida = int(cantidad_texto)
precio_convertido = float(precio_texto)

total_convertido = cantidad_convertida * precio_convertido

print(cantidad_convertida, type(cantidad_convertida))
print(precio_convertido, type(precio_convertido))
print(total_convertido)
```

`str()` convierte un número en texto. Esto puede ser útil para construir mensajes.

```python
numero_pedidos = 18
numero_pedidos_texto = str(numero_pedidos)

mensaje = "Número de pedidos: " + numero_pedidos_texto

print(mensaje)
print(type(numero_pedidos_texto))
```

## Salida con f-strings

Una f-string permite insertar variables dentro de un mensaje.

```python
producto = "Bowl Andino"
ingreso = 74000
margen = 39.46

resumen = (
    f"{producto} generó ${ingreso:,.0f} "
    f"con un margen de {margen:.2f}%."
)

print(resumen)
```

## Simulación de una entrada

`input()` devuelve texto. Para que este cuaderno pueda ejecutarse sin detenerse, se simula la entrada con valores de tipo `str`.

```python
dato_ingresado = "5"
precio_ingresado = "22000"

unidades = int(dato_ingresado)
precio = float(precio_ingresado)

total = unidades * precio

print(f"Unidades¿': {unidades}")
print(f"Precio: #$${precio:,.0f}")
print(f"Total: $&/{total:,.0f}")
```

## Lectura segura de un error de conversión

La palabra `"cuatro"` no puede convertirse directamente en un entero.

`try` y `except` permiten mostrar el error sin detener todo el cuaderno.

```python
valor_invalido = "cuatro"

try:
    numero = int(valor_invalido)
    print(numero)
except ValueError as error:
    print("No fue posible convertir el dato.")
    print("Tipo de error:", type(error).__name__)
```

# 2. Sesión 4 — Comparaciones y condiciones

Una comparación produce `True` o `False`.

```python
ventas = 820000
meta = 750000

print(ventas >= meta)
print(ventas == meta)
print(ventas != meta)
print(ventas < meta)
```

Los operadores lógicos combinan condiciones:

- `and`: ambas condiciones deben ser verdaderas;
- `or`: al menos una condición debe ser verdadera;
- `not`: invierte el resultado lógico.

```python
ventas = 820000
margen = 31.5

cumple_ventas = ventas >= 750000
cumple_margen = margen >= 30

print("Cumple ambas:", cumple_ventas and cumple_margen)
print("Cumple al menos una:", cumple_ventas or cumple_margen)
print("No cumple ventas:", not cumple_ventas)
```

## Condición simple con `if`

El bloque sangrado se ejecuta únicamente cuando la condición es verdadera.

```python
inventario = 8
inventario_minimo = 10

if inventario < inventario_minimo:
    print("Se debe solicitar reposición.")
```

## Selección doble con `if` y `else`

```python
ventas = 680000
meta = 750000

if ventas >= meta:
    print("La meta fue alcanzada.")
else:
    print("La meta todavía no fue alcanzada.")
```

## Selección múltiple con `if`, `elif` y `else`

```python
porcentaje_cumplimiento = 87

if porcentaje_cumplimiento >= 100:
    estado = "Meta cumplida"
elif porcentaje_cumplimiento >= 80:
    estado = "Avance satisfactorio"
else:
    estado = "Requiere seguimiento"

print(estado)
```

# 3. Sesión 5 — Validación y casos de prueba

Antes de realizar un cálculo se debe comprobar si los datos son válidos.

El siguiente ejemplo clasifica una cantidad como inválida, límite o normal.

```python
cantidad = -2

if cantidad < 0:
    print("Dato inválido: la cantidad no puede ser negativa.")
elif cantidad == 0:
    print("Caso límite: no se registraron unidades.")
else:
    print("Dato válido: puede continuar el cálculo.")
```

Ahora se ejecuta la misma lógica con tres valores diferentes. El código está repetido deliberadamente para observar los tres tipos de caso sin utilizar todavía ciclos ni funciones.

```python
cantidad = -1

if cantidad < 0:
    print("Caso 1:", "inválido")
elif cantidad == 0:
    print("Caso 1:", "límite")
else:
    print("Caso 1:", "normal")

cantidad = 0

if cantidad < 0:
    print("Caso 2:", "inválido")
elif cantidad == 0:
    print("Caso 2:", "límite")
else:
    print("Caso 2:", "normal")

cantidad = 5

if cantidad < 0:
    print("Caso 3:", "inválido")
elif cantidad == 0:
    print("Caso 3:", "límite")
else:
    print("Caso 3:", "normal")
```

## Condiciones anidadas

Una condición puede incluir otra condición dentro de su bloque.

```python
monto_compra = 180000
cliente_frecuente = True

if monto_compra >= 100000:
    if cliente_frecuente:
        descuento = 0.15
    else:
        descuento = 0.10
else:
    descuento = 0.00

valor_descuento = monto_compra * descuento
total_pagar = monto_compra - valor_descuento

print(f"Descuento: {descuento * 100:.0f}%")
print(f"Total a pagar: ${total_pagar:,.0f}")
```

# 4. Sesión 6 — Ciclos básicos

Un ciclo permite repetir instrucciones.

Para practicar `for` se utilizará una lista sencilla. El estudio formal de las colecciones se realizará más adelante.

```python
ventas_diarias = [120000, 95000, 143000, 87000, 165000,1]

for venta in ventas_diarias:
    print("Ventax registrada:", venta)
```

`range()` genera una secuencia de números.

```python
for numero in range(1, 10):
    print("Iteración:", numero)
```

## Acumulador

Un acumulador guarda una suma que crece durante el ciclo.

```python
ventas_diarias = [120000, 95000, 143000, 87000, 165000]
total_ventas = 0

for venta in ventas_diarias:
    total_ventas = total_ventas + venta

print("Total de ventas:", total_ventas)
```

## Contador

Un contador registra cuántas veces ocurre una condición.

```python
ventas_diarias = [120000, 95000, 143000, 87000, 165000]
ventas_superiores = 0

for venta in ventas_diarias:
    if venta >= 120000:
        ventas_superiores = ventas_superiores + 1

print("Días con ventas de al menos $120.000:", ventas_superiores)
```

## Ciclo `while`

`while` repite un bloque mientras la condición sea verdadera.

La variable de control debe actualizarse para evitar un ciclo infinito.

```python
saldo_meta = 100000
ahorro = 0
aporte_semanal = 25000
semana = 0

while ahorro < saldo_meta:
    semana = semana + 1
    ahorro = ahorro + aporte_semanal
    print(f"Semana {semana}: ahorro acumulado ${ahorro:,.0f}")

print("Meta alcanzada.")
```

# 5. Sesión 7 — Control y combinación de ciclos

`continue` omite el resto de la iteración actual.

En el siguiente ejemplo se ignoran los valores negativos.

```python
movimientos = [50000, -10000, 30000, -5000, 25000]
total_valido = 0

for movimiento in movimientos:
    if movimiento < 0:
        continue

    total_valido = total_valido + movimiento

print("Total de movimientos positivos:", total_valido)
```

`break` finaliza el ciclo cuando se cumple una condición.

```python
inventarios = [15, 12, 8, 20, 4, 17]
nivel_critico = 5

for inventario in inventarios:
    print("Inventario revisado:", inventario)

    if inventario <= nivel_critico:
        print("Se encontró un inventario crítico.")
        break
```

## Ciclos anidados

Un ciclo puede ejecutarse dentro de otro. Este recurso debe usarse con cuidado porque aumenta la cantidad de operaciones.

```python
sedes = ["Centro", "Norte"]
dias = ["Lunes", "Martes", "Miércoles"]

for sede in sedes:
    for dia in dias:
        print(sede, "-", dia)
```

## Búsqueda de un valor máximo

El ciclo compara cada elemento con el mayor valor conocido hasta ese momento.

```python
ventas_diarias = [120000, 95000, 143000, 87000, 165000]

mayor_venta = ventas_diarias[0]

for venta in ventas_diarias:
    if venta > mayor_venta:
        mayor_venta = venta

print("Mayor venta:", mayor_venta)
```

# 6. Sesión 8 — Funciones y modularidad

Una función agrupa instrucciones bajo un nombre.

`def` crea la función. Los parámetros reciben datos.

```python
def saludar(nombre):
    print(f"Hola, {nombre}. Bienvenido al curso.")

saludar("Andrea")
saludar("Carlos")
```

## Retorno de un valor

`return` entrega un resultado para que pueda almacenarse o utilizarse después.

```python
def calcular_total(cantidad, precio_unitario):
    total = cantidad * precio_unitario
    return total

total_venta = calcular_total(4, 18500)

print("Total calculado:", total_venta)
```

Imprimir y retornar no son lo mismo:

- `print()` muestra información;
- `return` devuelve un resultado al lugar donde se llamó la función.

```python
def mostrar_total(cantidad, precio):
    print(cantidad * precio)

def devolver_total(cantidad, precio):
    return cantidad * precio

resultado_mostrado = mostrar_total(2, 25000)
resultado_devuelto = devolver_total(2, 25000)

print("Valor de resultado_mostrado:", resultado_mostrado)
print("Valor de resultado_devuelto:", resultado_devuelto)
```

## Variables locales

Una variable creada dentro de una función existe únicamente durante la ejecución de esa función.

```python
def calcular_margen(ingreso, costo):
    margen_local = ingreso - costo
    return margen_local

margen = calcular_margen(74000, 44800)

print("Margen retornado:", margen)
print("La variable local se utilizó dentro de la función.")
```

## Función con decisión

Una función puede contener condiciones y retornar una clasificación.

```python
def clasificar_cumplimiento(porcentaje):
    if porcentaje >= 100:
        return "Meta cumplida"
    elif porcentaje >= 80:
        return "Avance satisfactorio"
    else:
        return "Requiere seguimiento"

print(clasificar_cumplimiento(105))
print(clasificar_cumplimiento(87))
print(clasificar_cumplimiento(62))
```

## Modularización

Un problema puede dividirse en funciones pequeñas, cada una con una responsabilidad clara.

```python
def calcular_ingreso(cantidad, precio_unitario):
    return cantidad * precio_unitario

def calcular_costo(cantidad, costo_unitario):
    return cantidad * costo_unitario

def calcular_margen(ingreso, costo):
    return ingreso - costo

cantidad = 4
precio_unitario = 18500
costo_unitario = 11200

ingreso = calcular_ingreso(cantidad, precio_unitario)
costo = calcular_costo(cantidad, costo_unitario)
margen = calcular_margen(ingreso, costo)

print(f"Ingreso: ${ingreso:,.0f}")
print(f"Costo: ${costo:,.0f}")
print(f"Margen: ${margen:,.0f}")
```

# 7. Ejemplo integrador

El siguiente programa reúne variables, conversiones, operaciones, condiciones, ciclos y funciones.

Ejecútelo completo y siga la secuencia de resultados.

```python
ventas_texto = ["120000", "95000", "143000", "87000", "165000"]
meta_total = 550000

def convertir_ventas(valores_texto):
    ventas_convertidas = []

    for valor in valores_texto:
        ventas_convertidas.append(float(valor))

    return ventas_convertidas

def calcular_total(valores):
    acumulado = 0

    for valor in valores:
        acumulado = acumulado + valor

    return acumulado

def evaluar_meta(total, meta):
    if total >= meta:
        return "Meta alcanzada"
    else:
        return "Meta pendiente"

ventas = convertir_ventas(ventas_texto)
total = calcular_total(ventas)
estado = evaluar_meta(total, meta_total)

print("Ventas convertidas:", ventas)
print(f"Total: ${total:,.0f}")
print("Estado:", estado)
```
