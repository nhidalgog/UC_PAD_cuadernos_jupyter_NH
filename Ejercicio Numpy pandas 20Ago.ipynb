##8. El mismo cálculo, dos formas
Escriba con NumPy lo que el ciclo hace con listas.


```python
# con listas de Python
lista = [24000, 28000, 22000, 30000, 8000]
resultado = []
for p in lista:
 resultado.append(p * 1.19)
print(resultado)
# con NumPy, en una sola línea
resultado_np = precios * 1.19
print(resultado_np)
```

##14. Dos condiciones a la vez

##18. Las funciones de resumen
Complete cada línea con el método que corresponde.


```python
ingresos = np.array([48000, 28000, 66000, 60000, 8000])
print(ingresos.sum()) # 210000
print(ingresos.mean()) # 42000.0
print(ingresos.max()) # 66000
print(round(ingresos.std(), 2))
```
