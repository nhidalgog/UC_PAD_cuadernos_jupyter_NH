Operaciones vectorizadas y broadcasting

```python
cantidades = np.array([2, 1, 3, 2, 1])
precios = np.array([24000, 28000, 22000, 30000, 8000])

ingresos = cantidades * precios     # [48000 28000 66000 60000 8000]
con_descuento = precios * 0.9       # el escalar alcanza a los cinco
total = ingresos.sum()              # 210000
print (total)
```

Ejercicio2

```python
precios = np.array([24000, 28000, 22000, 30000, 8000])

precios[0]        # 24000        primer elemento
precios[-1]       # 8000         último elemento
precios[1:4]      # [28000 22000 30000]   el final se excluye

mascara = precios > 20000     # [True True True True False]
precios[mascara]              # [24000 28000 22000 30000]
precios[mascara].size         # 4

np.where(precios > 25000, "alto", "normal")
# ['normal' 'alto' 'normal' 'alto' 'normal']
```

Ejercicio 3 con categoria

```python
precios = np.array([24000, 28000, 22000, 30000, 8000])

precios[0]        # 24000        primer elemento
precios[-1]       # 8000         último elemento
precios[1:4]      # [28000 22000 30000]   el final se excluye

mascara = precios > 20000     # [True True True True False]
precios[mascara]              # [24000 28000 22000 30000]
precios[mascara].size         # 4

categoria = np.where(precios > 25000, "alto", "normal")
# ['normal' 'alto' 'normal' 'alto' 'normal']
print (categoria.dtype)
print (categoria)
```

Ejercicio4 - Mascaras

```python
mascara_2 = (precios > 5000) & (precios < 26000) # & para "Y" |  para la condición "O"
print (mascara_2)
print (f"Resultado marcara #2: {mascara_2}")

mascara_3 = (precios > 5000) | (precios < 26000)  # & para "Y" |  para la condición "O"
print (mascara_2)
print (f"Resultado marcara #3: {mascara_3}")

```
